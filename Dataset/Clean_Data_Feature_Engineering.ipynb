```python
### Function to clean text:

### Basic text cleaning
def clean_text_basic(x):
    if pd.isna(x):
        return x
    x = str(x).lower().strip()
    x = re.sub(r"[^a-z0-9\s]", " ", x)
    x = re.sub(r"\s+", " ", x).strip()
    return x
```

The function clean_text_basic standardizes and cleans text data by first handling missing values, then converting the input to lowercase and trimming extra spaces. It removes all special characters and punctuation, keeping only letters, numbers, and spaces, and finally replaces multiple spaces with a single space to ensure consistency. Overall, it prepares raw text into a clean, uniform format suitable for analysis or modeling.

```python
df_clean2["arrival_mode_clean"] = df_clean2["arrival_mode"].apply(clean_text_basic)
```

```python
### Step2: Word Order Normalization
```

```python
def sort_words(x):
    if pd.isna(x):
        return x
    words = x.split()
    words_sorted = sorted(words)
    return " ".join(words_sorted)
```

The function sort_words standardizes text by first handling missing values, then splitting the input string into individual words, sorting those words alphabetically, and joining them back into a single string. This removes the effect of word order, so phrases with the same words but different arrangements (e.g., “chest pain severe” and “severe chest pain”) become identical, which can be useful for grouping or matching similar text entries.

```python
df_clean2["arrival_mode_sorted"] = df_clean2["arrival_mode_clean"].apply(sort_words)
```

```python
### Step 3: remove light stopwords
```

```python
### Remove stopwords (noise words)

def remove_stopwords(x):
    if pd.isna(x):
        return x
    words = x.split()
    words = [w for w in words if w not in stopwords]
    return " ".join(words)
```

```python
stopwords = {"and", "with", "from", "in", "to", "combined"}
```

```python
df_clean2["arrival_mode_tokens"] = df_clean2["arrival_mode_sorted"].apply(remove_stopwords)
```

```python
def categorize_arrival_mode(x):
    if pd.isna(x):
        return "Unknown"
    
    x = str(x).lower().strip()
    words = set(x.split())

    # explicit unknown
    if x in {"unknown", "unk", ""}:
        return "Unknown"

    # air transport first
    if any(w in words for w in [
        "air", "helicopter", "flight", "lifeflight", "stars", "sky", "nurse", "rn"
    ]):
        return "Air Transport"

    # police next
    if "police" in words:
        return "Police"

    # EMS / ambulance
    if any(w in words for w in [
        "ems", "ambulance", "municipal", "rural", "external", "local", "perimeter"
    ]):
        return "EMS"

    # stretcher
    if "stretcher" in words:
        return "Stretcher"

    # wheelchair and similar assisted mobility
    if any(w in words for w in [
        "wheelchair", "walker", "cane", "crutches", "stroller"
    ]):
        return "Wheelchair"

    # walk-in
    if any(w in words for w in [
        "walk", "walking"
    ]):
        return "Walk-in"

    # private / non-EMS transport
    if any(w in words for w in [
        "car", "taxi", "staxi", "bus"
    ]):
        return "Walk-in"

    # carried patients tend to be non-ambulatory
    if "carried" in words:
        return "Stretcher"

    return "Other"
```

**Notes**:

The Wait Time is the initial assessment minus the triage time:

**wait time = initial assessment time - triage time**

The length of Stay (LOS) is the last contact time minus the arrival time:

**LOS = last contact time - arrival time**

Thus, if there are some missing values in triage time, we can recover the missing values using wait time and initial assessment time. If both wait time and initial assessment time are missing, we will make triage time equal to arrival time.

Similarly, if there are missing values in the last contact time, we can recover them using LOS and arrival time.

Based on the formulas of wait time and LOS, we expect the following inequalities:

arrival time < triage time < initial assessment time < last contact time

wait time < LOS

In practice, these inequalities do NOT always hold, due to:

- Data entry errors
- System recording differences
- Missing + reconstructed values
- Workflow edge cases

Moreover, the above definition of wait time is correct if we define the wait time as the time from triage to physician assessment.
However, if we instead want the total waiting time from arrival, it would be: Wait time = initial assessment time − arrival time

```python
### Fix ED_complaint missing values and "UNKNOWN".

### Rename 'UNKNOWN' → 'unknown'
df_clean2['ED_complaint'] = df_clean2['ED_complaint'].replace({'UNKNOWN': 'unknown'})

### Fill NaN with 'unknown'
df_clean2['ED_complaint'] = df_clean2['ED_complaint'].fillna('unknown')
```

```python
def categorize_ed_complaint(x):
    if pd.isna(x):
        return "unknown"
    
    x = str(x).lower().strip()

    # normalize symbols
    x = x.replace("&", "and")

    # unknown stays as is
    if x in ["unknown", ""]:
        return "unknown"

    # ETRIAGE stays separate
    if "etriag" in x:
        return "ETRIAGE"

    # General
    if "general" in x or "minor" in x:
        return "General_Minor"

    # Cardiovascular
    if "cardio" in x:
        return "Cardiovascular"

    # Gastrointestinal
    if "gastro" in x:
        return "Gastrointestinal"

    # Orthopedic
    if "ortho" in x:
        return "Orthopedic"

    # Neurologic
    if "neuro" in x:
        return "Neurologic"

    # Respiratory
    if "respir" in x:
        return "Respiratory"

    # Mental Health
    if "mental" in x:
        return "Mental_Health"

    # Substance Misuse
    if "substance" in x:
        return "Substance_Misuse"

    # Trauma
    if "trauma" in x:
        return "Trauma"

    # OB-GYN
    if "obgyn" in x or "ob-gyn" in x:
        return "OBGYN"

    # ENT (all subtypes)
    if "ent" in x:
        return "ENT"

    return "Other"
```

```python
df_clean2["ED_complaint_final"] = df_clean2["ED_complaint"].apply(categorize_ed_complaint)
```

```python
def clean_chief_complaint(x):
    if pd.isna(x):
        return x
    x = str(x).lower().strip()
    x = re.sub(r"[^a-z0-9\s]", " ", x)
    x = re.sub(r"\s+", " ", x).strip()
    return x
```

```python
mapping = {
    "abd pain": "abdominal pain",
    "abdo pain": "abdominal pain",
    "sob": "shortness of breath",
    "cp": "chest pain",
    "n v": "nausea vomiting"
}
```

```python
### Step 3: Create Category Mapping Function
```

```python
def map_chief_complaint(x):
    if pd.isna(x):
        return "other"
    
    # cardiac
    if "chest pain" in x:
        return "cardiac"
    
    # respiratory
    if any(k in x for k in ["breath", "cough", "asthma"]):
        return "respiratory"
    
    # GI
    if any(k in x for k in ["abdominal", "nausea", "vomit", "diarrhea"]):
        return "gastrointestinal"
    
    # neuro
    if any(k in x for k in ["headache", "dizzy", "seizure", "syncope"]):
        return "neurological"
    
    # trauma
    if any(k in x for k in ["fall", "injury", "fracture", "trauma"]):
        return "trauma"
    
    # infection
    if "fever" in x:
        return "infection"
    
    # musculoskeletal
    if any(k in x for k in ["back pain", "joint", "leg pain"]):
        return "musculoskeletal"
    
    # skin
    if any(k in x for k in ["rash", "wound", "cut"]):
        return "skin"
    
    # mental
    if any(k in x for k in ["anxiety", "depression"]):
        return "mental_health"
    
    # general
    if any(k in x for k in ["weakness", "fatigue", "unwell"]):
        return "general"
    
    # unknown
    if x == "unknown":
        return "unknown"
    
    return "other"
```

```python
### Period of Day
# Morning: 6:00 - 11:59, Afternoon: 12:00 - 17:59, Evening: 18:00 - 21:59, Night: 22:00 - 5:59

def get_period_of_day(hour):
    if 6 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 18:
        return 'Afternoon'
    elif 18 <= hour < 22:
        return 'Evening'
    else:
        return 'Night'
```

```python
df_clean3['period_of_day'] = df_clean3['arrival_time'].dt.hour.map(get_period_of_day)
```

```python
### Day of Week
### e.g. 'Monday', 'Tuesday'

df_clean3['day_of_week'] = df_clean3['arrival_time'].dt.day_name()  
```

```python
### Season (Manitoba - Northern Hemisphere) ---
### Winter: Dec-Feb, Spring: Mar-May, Summer: Jun-Aug, Fall: Sep-Nov

def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    else:
        return 'Fall'
```

```python
df_clean3['season'] = df_clean3['arrival_time'].dt.month.map(get_season)
```

```python
### Holiday (Manitoba)
### Get the range of years in the data to cover all holidays

years = df_clean3['arrival_time'].dt.year.dropna().unique().astype(int).tolist()
mb_holidays = holidays.Canada(prov='MB', years=years)
```

```python
df_clean3['holiday'] = df_clean3['arrival_time'].dt.date.map(
    lambda x: 'Holiday' if x in mb_holidays else 'Not Holiday'
)
```

```python
### Classification Logic

admitted_categories = [
    'ED TRSF to IP',
    'ED TRSF to IP Critical Care/OR',
    'TRSF to Acute Care',
    'ED TRSF to Internal ER',
    'ED TRSF to Ext Acute Care/ER',
    'TRSF to Other Facility',

    'ED TRSF to Internal Clinic',
    'ED TRSF to Internal DNC',
    'ED TRSF to Ext Non Acute Care',
    'TRSF to Continuing Care'
]

not_admitted_categories = [
    'ED Home',
    'Home',
    'Home with Support Services',
    'ED Home with Support Services',
    'ED Reg/Triaged LWBS',
    'ED Reg LWBS',
    'ED Reg/Triaged/Assess LWBS',
    'ED Reg/Triaged/Assess LAMA',
    'Left Against Medical Advice',
    'No Return from Pass'
]
```
